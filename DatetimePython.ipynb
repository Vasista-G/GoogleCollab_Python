```python
from datetime import date

d = date(2026,7,29)

print(d)

#accesing individual values
print(d.day, d.month, d.year)

#get current date

date = date.today()

print(f"Today_file_{date.today()}")
```

```python
from datetime import time

t = time(10,20,33)
print(t)
print(t.hour)
```

```python
from datetime import datetime

print(datetime.today())
```

Creating DateTime from string


```python
date_string = "18-08-2026 2:30 AM"

print(type(date_string))

strtotime = datetime.strptime(date_string, "%d-%m-%Y %I:%M %p")

print(strtotime)
```

```python
from datetime import timedelta, date

f = date.today() - timedelta(days=10)

print(f)

```

```python
#Finding a weekday - we can tell you which day of the week a date falls on.

'''
weekday() returns:

Monday    = 0
Tuesday   = 1
Wednesday = 2
Thursday  = 3
Friday    = 4
Saturday  = 5
Sunday    = 6

'''

from datetime import date

d = date(2026, 8, 18)

weekday = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print(d.weekday())
print(weekday[d.weekday()])
```

Comparing datetime objects

```python
from datetime import datetime

dt1 = datetime(2026, 8, 18, 10, 30)
dt2 = datetime(2026, 8, 18, 15, 30)

if dt1 < dt2:
    print("dt1 occurs before dt2")
```

A practical example: appointment system

Imagine you're building an appointment system.

```python
from datetime import datetime, timedelta

dt = datetime(day = 20, month = 8, year = 2026, hour= 5, minute=30)

print("Appointment on :", dt)

remainder = dt - timedelta(hours = 2)

print("Remainder at :", remainder)
```

```python
# Finding Expire data - Suppose a product expires 30 days after buying

purchase_date = datetime.now()

expire_date = purchase_date + timedelta(days = 30)

print(f"Purchase date {purchase_date} - Expire Date {expire_date}")
```
