```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Owen Harris", "Henry", "Elizabeth", "Mickey"],
    "Age": [22, 35, 58, 22],
    "Gender": ["male", "male", "female", "male"]
})

df
```

```python
emp = pd.read_csv("employee.csv")

emp.head()
```

```python
 (
emp.loc[emp["department"] == "IT", ["name", "salary","department"]]
.sort_values(by = "salary", ascending = False)
.nlargest(3, "salary")
)
```

```python
#Find employees who work in IT OR Finance

emp[(emp["department"] == "IT") | (emp["department"] == "Finance")]
```

```python
#2.	Find employees who are from Hyderabad or Bangalore AND have a salary above 60,000

emp.loc[((emp["city"] == "Hyderabad") | (emp["city"] == "Bangalore")) & (emp["salary"] > 60000)]
```

```python
'''
3.	Find employees who have:
	More than 5 years of experience
	Performance rating >= 4
	Salary > 70,000

'''

(
    emp.loc
    [
        (emp["experience_years"] > 5.0) &
        (emp["performance_rating"] >= 4) &
        (emp["salary"] > 70000),
        ["employee_id", "name", "salary"]

    ]
)
```

```python
#4.	Find employees whose salary is between 60,000 and 80,000, inclusive.

emp[emp["salary"].between(60000, 80000)]
```

```python
#5.	Find employees whose name starts with "A".

emp[emp["name"].str.startswith("A")][["employee_id", "name"]]

emp.loc[emp["name"].str.startswith("A"), ["employee_id", "name","department"]]
```

```python
#6.	Find employees whose name contains the letter "a", regardless of capitalization.

emp.loc[emp["name"].str.contains("a", case = False), ["employee_id", "name"]]
```

```python
#9.	Using .loc[], increase the salary of all IT employees by 10%.
# assign function does not modify original df. it creates a new df

emp.loc[emp["department"] == "IT", ["name","salary"]].assign(new_salary = lambda x : x["salary"] * (110/100))
```

```python
#10. Change the performance rating of employees with salary greater than 90,000 to 5.

#emp.loc[emp["salary"] > 90000, "performance_rating"] = 5

emp[emp["salary"] > 90000]
```

```python
#SECTION 3 - VALUE_COUNTS()

#11. Find how many employees belong to each department.

emp["department"].value_counts()
```

```python
#14. Find the percentage of employees belonging to each department.

emp["department"].value_counts(normalize=True) * 100
```
