```python
emp = pd.read_csv("employee.csv")

#emp.head()
```

# **Loc and iLoc and Conditional filtering**

```python
 (
emp.loc[emp["department"] == "IT", ["name", "salary","department"]]
.sort_values(by = "salary", ascending = False)
.nlargest(3, "salary")
)
```

```python
#Find employees who work in IT OR Finance

emp[(emp["department"] == "IT") | (emp["department"] == "Finance")]
```

```python
#2.	Find employees who are from Hyderabad or Bangalore AND have a salary above 60,000

emp.loc[((emp["city"] == "Hyderabad") | (emp["city"] == "Bangalore")) & (emp["salary"] > 60000)]
```

```python
'''
3.	Find employees who have:
	More than 5 years of experience
	Performance rating >= 4
	Salary > 70,000

'''

(
    emp.loc
    [
        (emp["experience_years"] > 5.0) &
        (emp["performance_rating"] >= 4) &
        (emp["salary"] > 70000),
        ["employee_id", "name", "salary"]

    ]
)
```

```python
#4.	Find employees whose salary is between 60,000 and 80,000, inclusive.

emp[emp["salary"].between(60000, 80000)]
```

```python
#5.	Find employees whose name starts with "A".

emp[emp["name"].str.startswith("A")][["employee_id", "name"]]

emp.loc[emp["name"].str.startswith("A"), ["employee_id", "name","department"]]
```

```python
#6.	Find employees whose name contains the letter "a", regardless of capitalization.

emp.loc[emp["name"].str.contains("a", case = False), ["employee_id", "name"]]
```

```python
#9.	Using .loc[], increase the salary of all IT employees by 10%.
# assign function does not modify original df. it creates a new df

emp.loc[emp["department"] == "IT", ["name","salary"]].assign(new_salary = lambda x : x["salary"] * (110/100))
```

```python
#10. Change the performance rating of employees with salary greater than 90,000 to 5.

#emp.loc[emp["salary"] > 90000, "performance_rating"] = 5

emp[emp["salary"] > 90000]
```

# **Section 4 GROUP BY**

```python
#GROUP BY

#15.Find the average salary for each department.

emp.groupby("department")["salary"].mean()
```

```python
#16.Find the maximum salary for each city.

emp.groupby("city", as_index= False)["salary"].max()
```

```python
#17.	Find the minimum and maximum salary for each department.

emp.groupby("department")["salary"].agg(max_salary = "max", min_sal = "min") # to give names to aggr columns

# Below we are specifying aggregate column in the agg function not after groupby like above
emp.groupby("department").agg(max_sal = ("salary", "max"), min_sal = ("salary","min"))

#to perform multiple aggregations on a column
emp.groupby("department")["salary"].agg(["max","min"])

#Using multiple functions for multiple columns
emp.groupby("department").agg({"salary": ["max", "min"], "experience_years": "mean"})

```

```python
#18.	Find the average salary and average experience for each department.

emp.groupby("department").agg({"salary": "mean", "experience_years": "mean"})
```

```python
#19.	Find the number of employees and average salary for each city.

#emp.groupby("city").agg(emp_Count = ("employee_id","count"), avg_sal = ("salary", "mean"))
```

```python
#20.	Find the average salary of employees for each performance rating.

emp.groupby("performance_rating")["salary"].mean().round(2).reset_index()
```

# **SECTION 5 - MULTIPLE AGGREGATIONS**

```python
'''
For each department, calculate:

Number of employees

Average salary

Maximum salary

Minimum salary

Average experience

'''

(
    emp.groupby("department")
    .agg
     (
        count_emp = ("employee_id","count")
        , avg_sal = ("salary","mean")
        , Max_sal = ("salary", "max")
        , Min_sal = ("salary","min")
       , avg_exp = ("experience_years","mean")
     )
)
```

# **SECTION 6 - TRANSFORM()**

```python
#Create a new column called: department_avg_salary

emp["dept_avg_sal"] = emp.groupby("department")["salary"].transform("mean")

#Create a new column called: salary_difference It should contain: Employee Salary - Department Average Salary

emp["sal_difference"] = (emp["salary"] - emp["dept_avg_sal"]).abs()

emp[["employee_id", "name", "salary", "sal_difference"]]
```

```python
'''
Create a new column called: above_department_average

The column should contain True or False depending on whether the
employee's salary is above their department's average salary.
'''

emp.loc[emp["salary"] > emp["dept_avg_sal"]]

```

# **SECTION 7 - RANKING**

```python
#Create a salary_rank column that ranks employees from highest salary to lowest salary.

emp["salary_rank"] = emp["salary"].rank(ascending= False, method = "min")

emp[["salary_rank", "salary"]].sort_values(by = "salary", ascending = False)
```

```python
#Find the top 5 employees based on salary.

emp[emp["salary_rank"] <= 5][["employee_id", "salary", "salary_rank"]].sort_values(by = "salary_rank")
```

```python
#Find the top 2 highest-paid employees from each department.

emp.sort_values(by = ["department","salary"], ascending= False).groupby("department").head(2)
```
