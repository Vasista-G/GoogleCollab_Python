```python
df["Price"] = 5

def NewPrice(row):
    if row["Coffee Type"] == "Espresso":
        return 5
    else:
        return 4

df["New_price"] = df.apply(NewPrice, axis = 1)

df["Total Price"] = df["New_price"] * df["Units Sold"]

#df["New_price"] = df["Coffee Type"].apply(lambda row : 4 if row == "Espresso" else 2)

df
```
