Writing to a txt

```python
#Writing to a Txt
file_path = "pract.txt"

with open(file = file_path, mode = "w") as file:
    file.write("Hello, World!")

#Using Data Structures
emp_names = ["S", "D", "Y", "N"]

with open(file = file_path, mode = "w") as file:
    for name in emp_names:
        file.write(name + "\n")


#Using Dict

emp_info = {"name": "S",
            "age": 25,
            "dept": "IT"}

with open(file = file_path, mode = "w") as file:
    for key, value in emp_info.items():
        file.write(f"{key}: {value}\n")


#Writing Multiple Lines Using writelines()

emp_names = ["python\n", "is powerful \n", "language"]

with open(file = file_path, mode = "w") as file:
    file.writelines(emp_names)


```

Writing to a json file

```python
import json

#simple json
student = {
    "name": "John",
    "age": 20,
    "grade": "A"
}

with open("student.json", "w") as file:
    json.dump(student, file)

#Nested Json Objects

student = {
    "student": {
        "name": "John",
        "age": 20,
        "grade": "A"
    }
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)


#Arrays - Python lists become JSON arrays.
students = {
    "students": [
        {
            "name": "John",
            "grade": "A"
        },
        {
            "name": "Alice",
            "grade": "B"
        }
    ]
}

with open("students.json", "w") as file:
    json.dump(students, file, indent=4)



#Creating Dynamic Json Content
name = "John"
age = 20
subjects = ["Python", "Java"]

student = {}

student["student"] = {}


student["student"]["name"] = name
student["student"]["age"] = age
student["student"]["subjects"] = subjects

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)
```

```python
import json

#Creating Dynamic Json Content
name = ["John","Hold", "Gold"]
age = 20
subjects = ["Python", "Java"]

student = {}

student["student"] = {}

print(student)

studentid = [1,2,3]

zip_obj = zip(studentid, name)

for id, name in zip_obj:
    student["student"][id] = {}
    student["student"][id]["name"] = name
    student["student"][id]["age"] = age
    student["student"][id]["subjects"] = subjects

print(student)

with open("student.json", "w") as file:
    json.dump(student, file, indent = 4)


```

```python
import json

names = ["John","Hold", "Gold"]
age = 20
subjects = ["Python", "Java"]
studentid = [1,2,3]

#Dict Comprehension
student = {
    "student" : {
        id : {
            "name" : name,
            "age" : age,
            "subjects" : subjects
        } for id, name in zip(studentid, names)
    }
}

print(student)
```

Writing to a CSV

```python
import csv

student = [
    ["Name", "Age", "Grade"],
    ["John", 20, "A"],
    ["Alice", 22, "B"],
    ["Bob", 19, "C"]
]

with open("student.csv", "w", newline = "") as file:
    writer = csv.writer(file)
    writer.writerows(student)
```

```python
import csv

#Using Dictinaries

students = [
    {"Name": "Alice", "Math": 95, "Science": 90},
    {"Name": "Bob", "Math": 82, "Science": 88},
    {"Name": "Charlie", "Math": 91, "Science": 94}
]

with open("marks.csv", "w", newline="") as file:

    fields = ["Name", "Math", "Science"]

    writer = csv.DictWriter(file, fieldnames=fields)

    writer.writeheader()

    writer.writerows(students)
```
